# 산업재해 사망사고 데이터 분석

## 1. 데이터 불러오기 및 기본 확인

원본 산업재해 CSV 데이터를 불러오고 데이터 크기와 기본 구조를 확인합니다.

```python
import pandas as pd

file_path = "../data/raw/industrial_accident_deaths.csv"
df = pd.read_csv(file_path, encoding="cp949")
df.head()
```

```python
print("데이터 크기:", df.shape)
print("컬럼 목록:", df.columns.tolist())
```

## 2. 데이터 품질 확인

결측치, 중복 행, 데이터 자료형을 확인하여 분석에 사용할 데이터의 기본 품질을 점검합니다.

```python
print("전체 결측치:", df.isnull().sum().sum())
print("중복 행:", df.duplicated().sum())
df.info()
```

## 3. 업종 및 사업장 규모 구조 확인

대업종, 중업종, 사업장 규모의 분포를 확인하여 데이터가 어떤 기준으로 구성되어 있는지 파악합니다.

```python
print("대업종 개수:", df["대업종"].nunique())
print("중업종 개수:", df["중업종"].nunique())
print("\n규모 종류:")
print(df["규모"].value_counts())
```

## 4. 총계 데이터 검증

사고 유형별 사망자 수의 합계와 `총계` 컬럼이 일치하는지 확인하여 데이터의 계산 오류 여부를 검증합니다.

```python
accident_cols = df.columns[4:]  # 떨어짐부터 분류불능까지
calculated_total = df[accident_cols].sum(axis=1)
mismatch = df["총계"] != calculated_total

print("총계 불일치 행:", mismatch.sum())
df.loc[mismatch, ["대업종", "중업종", "규모", "총계"]].head()
```

## 5. 분석용 데이터 저장

검증이 완료된 데이터를 UTF-8 형식의 CSV 파일로 저장하여 이후 분석 노트북에서 사용할 수 있도록 준비합니다.

```python
output_path = "../data/processed/industrial_accident_deaths_clean.csv"
df.to_csv(output_path, index=False, encoding="utf-8-sig")
print("분석용 CSV 저장 완료:", output_path)
```
